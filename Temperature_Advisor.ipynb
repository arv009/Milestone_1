```python
def temperature_today(temperature): # Gets the temperature today
  
  if temperature > 50:
    return "Temperature is too high. Double check the temperature and input the temperature again."
  elif temperature >= 35:
    return f"It's a hot day today! The temperature is {temperature} degrees Celsius. Wear light clothing and stay hydrated."
  elif temperature >= 25:
    return f"It's a warm day today! The temperature is {temperature} degrees Celsius. Wear light clothing."
  elif temperature >= 15:
    return f"It's a cool day today! The temperature is {temperature} degrees Celsius. Wear a jacket or sweater."
  elif temperature >= 5: 
    return f"It's a cold day today! The temperature is {temperature} degrees Celsius. Wear warm clothing and stay indoors if possible."
  else:
    return f"It's very cold today! The temperature is {temperature} degrees Celsius. Wear winter clothing and stay indoors if possible."

while True: # The while loop will keep running until it hits 'break'
    temperature_input = input("Please enter today's temperature in Celsius: ") # Get the temperature using input

    try:
      current_temperature = float(temperature_input) # Convert the input to a float
      print (temperature_today(current_temperature))
      
      if current_temperature <= 50:
        break # Only exit when temperature is less than or equal to 50
    except ValueError:
      print (f"You entered '{temperature_input}' which is not a valid number. Please enter a valid number for the temperature.") # Handles the error when the input is not a valid number
```

```python
assert temperature_today(36.8) == "It's a hot day today! The temperature is 36.8 degrees Celsius. Wear light clothing and stay hydrated."
assert temperature_today(28.0) == "It's a warm day today! The temperature is 28.0 degrees Celsius. Wear light clothing."
assert temperature_today(20.1) == "It's a cool day today! The temperature is 20.1 degrees Celsius. Wear a jacket or sweater."
assert temperature_today(12.6) == "It's a cold day today! The temperature is 12.6 degrees Celsius. Wear warm clothing and stay indoors if possible."
print("All checks passed!")
```
